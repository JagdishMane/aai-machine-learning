# How to load datasets with Seaborn


```python
import seaborn as sb
```

```python
sb.get_dataset_names()
```

# Seaborn plot
**seaborn.plot(data, x, y, hue)**

- data = dataset
- x - Variables for x axis
- y - Variables for y axis
- hue - Grouping variable.

```python
data = sb.load_dataset('titanic')
data.head()
```

```python
# Count plot
sb.countplot(data=data, x='sex')
```

```python
sb.countplot(data=data, y='sex')
```

```python
sb.countplot(data=data, x='sex', hue='alive')
```

```python
## Continous plots
data = sb.load_dataset('exercise')
data.head()
```

```python
data = sb.load_dataset('diamonds')
data.head()
```

```python
data = data.sample(3000)
```

```python
sb.scatterplot(data=data, x="carat", y="price")
```

```python
sb.scatterplot(data=data, x="carat", y="price", marker="+")
```

```python
sb.kdeplot(data=data, x="carat", y="price")
```

```python
sb.histplot(data=data, x="carat", y="price")
```

```python
sb.kdeplot(data=data, x="carat", y="price", cmap='coolwarm')
sb.histplot(data=data, x="carat", y="price", color='red')
```

```python
sb.jointplot(data=data, x="carat", y="price")
```

```python
sb.jointplot(data=data, x="carat", y="price", kind='hex')
```

```python
sb.lmplot(data=data, x="carat", y="price", markers=".", line_kws={"color": "red"})
## Regression plot
```

## Plotting relationship


```python
sb.relplot(data=data, x='carat', y='price', col='cut')
```

# Correlations

```python
data=sb.load_dataset('diamonds')
data.head()
```

```python
sb.heatmap(data.select_dtypes(include='number').corr())
```

```python
sb.heatmap(data.select_dtypes(include='number').corr(), annot=True, linewidths=1)
```

```python
sb.heatmap(data.select_dtypes(include='number').corr()[['price']], annot=True, linewidths=1)
```
