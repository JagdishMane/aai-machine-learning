# Assignment 1 - EDA on Home Credit Default Risk dataset

```python
### Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

## Seaborn Theme
sns.set_theme()
```

```python
### Load dataset
df = pd.read_csv('train_data.csv')
df.head()
```

```python
### Print shape
df.shape
```

```python
# Describe the dataset
df.describe()
```

### Check Column Names

```python
print("Number of columns:", len(df.columns))
display(df.columns.tolist())
```

### Check Missing Values

```python
missing_values_count = df.isnull().sum()
missing_values_percent = (df.isnull().sum() / len(df)) * 100

missing_info = pd.DataFrame({
    'Missing Count': missing_values_count,
    'Missing Percentage': missing_values_percent
})

missing_info = missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)

print("Columns with Missing Values:")
display(missing_info)
```

```python
### ased on the missing_info DataFrame, let's drop the columns with a large percentage of missing values,
### specifically those with more than 50% missing data. This will help simplify the dataset and potentially improve model performance

cols_to_drop = missing_info[missing_info['Missing Percentage'] > 50].index.tolist()
df = df.drop(columns=cols_to_drop)

print(f"Dropped {len(cols_to_drop)} columns with more than 50% missing values.")
print(f"New DataFrame shape: {df.shape}")
display(df.head())
```

### Understand the Target Variable

```python
plt.figure(figsize=(7, 5))
sns.countplot(x='TARGET', data=df, palette='viridis')
plt.title('Distribution of Target Variable')
plt.xlabel('TARGET (0: Repaid, 1: Defaulted)')
plt.ylabel('Count')
plt.xticks([0, 1], ['Repaid (0)', 'Defaulted (1)'])
plt.show()

print('Value Counts for TARGET:')
display(df['TARGET'].value_counts())
print('\nPercentage of each class for TARGET:')
display(df['TARGET'].value_counts(normalize=True) * 100)
```
