#  Data Analysis

### Handle missing values


```python
missing_values_count = df.isnull().sum()
missing_values_percent = (df.isnull().sum() / len(df)) * 100

missing_info = pd.DataFrame({
    'Missing Count': missing_values_count,
    'Missing Percentage': missing_values_percent
})

missing_info = missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)

print("Columns with Missing Values:")
display(missing_info)
```

```python
### Based on the missing_info DataFrame, let's drop the columns with a large percentage of missing values,
### specifically those with more than 50% missing data. This will help simplify the dataset and potentially improve model performance

cols_to_drop = missing_info[missing_info['Missing Percentage'] > 50].index.tolist()
df = df.drop(columns=cols_to_drop)

print(f"Dropped {len(cols_to_drop)} columns with more than 50% missing values.")
print(f"New DataFrame shape: {df.shape}")
display(df.head())
```

# Visualization

### Countplot on Target Variable

```python
sns.countplot(x='TARGET', data=df, palette='viridis')
```

- The countplot shows there are around ~100000 loan defaulters and 140000 repaid.
- Defaulter are less compared to the Repaid, ie 90% of client repaid the loans and around 10% defaulted.
- Hence there is class imbalance in the dataset as repaid count is very high, so this could impact the model prediction.

## Histplot -  Annual total Income vs Target

```python
#sns.histplot(x='AMT_INCOME_TOTAL', data=df, hue='TARGET')
plt.figure(figsize=(10,5))

# Remove extreme outliers using 99th percentile
income_limit = df['AMT_INCOME_TOTAL'].quantile(0.99)

sns.histplot(
    data=df[df['AMT_INCOME_TOTAL'] <= income_limit],
    x='AMT_INCOME_TOTAL',
    hue='TARGET',
    bins=40,
    kde=True
)

plt.title("Income Distribution vs TARGET")
plt.xlabel("Total Income")
plt.ylabel("Count")

plt.show()
```

- Income ranges are heavility concentrated between 50000 to 200000.
- Income peak is around 130000 to 150000


## Boxplot - AMT_INCOME_TOTAL vs TARGET

```python
sns.boxplot(data=df, x='TARGET', y='AMT_INCOME_TOTAL')
plt.ylim(0, 400000)
plt.show()
```

- Median income for defaults (~135K) is lower than paid clients ~150k.
- However many defaults earn same as repairs.
- Repairs outliers are high.

### Correlation

```python
numeric_df = df.select_dtypes(include=np.number)
correlations = numeric_df.corr()['TARGET'].sort_values()
print("Top Positive Correlations:\n", correlations.tail(10))
print("\nTop Negative Correlations:\n", correlations.head(10))
```

#### Negative Correlation
- EXT_SOURCE_3 & EXT_SOURCE_2 (-0.17 to -0.15): These are your strongest features. A higher score here strongly suggests a lower risk of default. They are likely credit scores from external bureaus.
- DAYS_EMPLOYED : Suggests that people who have been at their jobs longer (more negative numbers) are safer borrowers.

### Positive Correlation

- DAYS_BIRTH: indicates that younger applicants are statistically riskier
- REGION_RATING_CLIENT_W_CITY indicates worse performing regions indicates a higher risk of defaults.
- DAYS_LAST_PHONE_CHANGE People who changed their phone very recently  appear to be slightly higher risk

```python


plt.figure(figsize=(10, 6))
# KDE plot shows the 'shape' of the data for each group
sns.kdeplot(df.loc[df['TARGET'] == 0, 'EXT_SOURCE_3'], label='Target == 0 (Repayers)', fill=True)
sns.kdeplot(df.loc[df['TARGET'] == 1, 'EXT_SOURCE_3'], label='Target == 1 (Defaulters)', fill=True)

plt.title('Distribution of EXT_SOURCE_3 by Target Status')
plt.xlabel('EXT_SOURCE_3 Score')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
# Convert negative days to positive years
df['AGE'] = df['DAYS_BIRTH'] / -365

plt.figure(figsize=(10, 6))
sns.kdeplot(df.loc[df['TARGET'] == 0, 'AGE'], label='Target == 0 (Repayers)', shade=True)
sns.kdeplot(df.loc[df['TARGET'] == 1, 'AGE'], label='Target == 1 (Defaulters)', shade=True)

plt.title('Distribution of Age by Target Status')
plt.xlabel('Age (Years)')
plt.legend()
plt.show()
```
